## 7. Part F - Basic OOP for data analysis

The class below stores the dataset and provides reusable methods for common analysis tasks.

```python
class IrisAnalyzer:
    def __init__(self, data):
        self.data = data

    def show_shape(self):
        return self.data.shape

    def species_counts(self):
        return self.data["species"].value_counts()

    def feature_mean_by_species(self, feature):
        return self.data.groupby("species")[feature].mean()

    def feature_max_by_species(self, feature):
        return self.data.groupby("species")[feature].max()

    def filter_by_species(self, species_name):
        return self.data[self.data["species"] == species_name]

    def plot_feature_by_species(self, feature):
        sns.boxplot(data=self.data, x="species", y=feature)
        plt.title(f"{feature} by species")
        plt.show()

analyzer = IrisAnalyzer(df)
print(analyzer.show_shape())
print(analyzer.species_counts())
print(analyzer.feature_mean_by_species("petal_length"))
print(analyzer.feature_max_by_species("petal_length"))
analyzer.plot_feature_by_species("sepal_length")
```

**Task F1:** Add a method called `feature_max_by_species(self, feature)`.

**Task F2:** Add a method called `filter_by_species(self, species_name)`.

**Task F3:** Use the class to plot `sepal_length` by species.

```python
class IrisAnalyzerV2:
    def __init__(self, data):
        self.data = data.copy()

    def check_feature(self, feature):
        if feature not in self.data.columns:
            raise ValueError(f"Feature {feature} does not exist in the DataFrame.")

    def basic_summary(self):
        return self.data.describe()

    def mean_by_species(self, feature):
        self.check_feature(feature)
        return self.data.groupby("species")[feature].mean()

    def max_by_species(self, feature):
        self.check_feature(feature)
        return self.data.groupby("species")[feature].max()

    def min_by_species(self, feature):
        self.check_feature(feature)
        return self.data.groupby("species")[feature].min()

    def plot_boxplot(self, feature):
        self.check_feature(feature)
        sns.boxplot(data=self.data, x="species", y=feature)
        plt.title(f"{feature} by species")
        plt.show()

    def plot_histogram(self, feature):
        self.check_feature(feature)
        plt.figure(figsize=(6, 4))
        plt.hist(self.data[feature], bins=12)
        plt.xlabel(feature)
        plt.ylabel("Frequency")
        plt.title(f"Distribution of {feature}")
        plt.show()

analyzer2 = IrisAnalyzerV2(df)
print(analyzer2.mean_by_species("petal_area"))
print(analyzer2.min_by_species("petal_area"))
analyzer2.plot_histogram("petal_area")
```
