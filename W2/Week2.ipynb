```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("Lec_2.1_train.csv")

print("Iris dataset loaded successfully!")
display(df.head())
```

```python
print("Columns:")
print(df.columns)
```

```python
print("Missing values:")
print(df.isnull().sum())
```

```python
print("Descriptive statistics:")
display(df.describe())
```

```python
print("Descriptive statistics:")
display(df.describe())
```

```python
# Calculate sepal area
df["sepal_area"] = (
    df["sepal length (cm)"] *
    df["sepal width (cm)"]
)

# Calculate petal area
df["petal_area"] = (
    df["petal length (cm)"] *
    df["petal width (cm)"]
)

print("New derived columns created:")
display(df.head())
```

```python
# Flowers with petal length greater than 5 cm
df_long_petals = df[
    df["petal length (cm)"] > 5
]

print("Flowers with petal length greater than 5 cm:")
display(df_long_petals)
```

```python
# Flowers with sepal width greater than 3 cm
df_wide_sepals = df[
    df["sepal width (cm)"] > 3
]

print("Flowers with sepal width greater than 3 cm:")
display(df_wide_sepals)
```

```python
# Only Virginica flowers
df_virginica = df[
    df["species"] == "virginica"
]

print("Virginica flowers:")
display(df_virginica)
```

```python
# Compare average sepal length and petal length by species
species_comparison = df.groupby("species")[
    ["sepal length (cm)", "petal length (cm)"]
].mean()

print("Average measurements by species:")
display(species_comparison)
```

```python
plt.figure(figsize=(8, 5))

for species in df["species"].unique():
    data = df[df["species"] == species]

    plt.scatter(
        data["sepal length (cm)"],
        data["petal length (cm)"],
        label=species
    )

plt.xlabel("Sepal Length (cm)")
plt.ylabel("Petal Length (cm)")
plt.title("Sepal Length vs Petal Length")
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(
    df["petal length (cm)"],
    bins=10
)

plt.xlabel("Petal Length (cm)")
plt.ylabel("Frequency")
plt.title("Distribution of Petal Length")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="species",
    y="petal length (cm)"
)

plt.xlabel("Species")
plt.ylabel("Petal Length (cm)")
plt.title("Petal Length by Species")
plt.show()
```

```python
plt.figure(figsize=(8, 6))

correlation = df.drop(
    columns=["species"]
).corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm"
)

plt.title("Feature Correlation Heatmap")
plt.show()
```

```python
class IrisAnalyzer:

    def __init__(self, dataframe):
        self.dataframe = dataframe

    # Method 1
    def show_first_rows(self, n=5):
        return self.dataframe.head(n)

    # Method 2
    def show_statistics(self):
        return self.dataframe.describe()

    # Method 3
    def average_by_species(self):
        return self.dataframe.groupby("species")[
            ["sepal length (cm)", "petal length (cm)"]
        ].mean()

    # Method 4
    def filter_species(self, species):
        return self.dataframe[
            self.dataframe["species"] == species
        ]
```

```python
# Create an IrisAnalyzer object
analyzer = IrisAnalyzer(df)

print("First 5 rows:")
display(analyzer.show_first_rows())

print("Statistics:")
display(analyzer.show_statistics())

print("Average measurements by species:")
display(analyzer.average_by_species())

print("Setosa flowers:")
display(analyzer.filter_species("setosa"))
```

The most important pattern in the Iris dataset is that the flower species have different measurements, especially in petal length and petal width. The scatter plot and boxplot show that petal measurements can help distinguish the different species. The heatmap also shows relationships between the different flower features.
